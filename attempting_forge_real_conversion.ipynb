```python
def get_experimenter(subject_fullname):
    import datajoint as dj
    subject = dj.create_virtual_module("subject", "u19_subject")
    lab = dj.create_virtual_module("lab", "u19_lab")
    exp = (subject.Subject() * lab.User() & f"subject_fullname = '{subject_fullname}'").fetch("full_name", limit=1)
    experimenter = exp[0]
    return experimenter
```
